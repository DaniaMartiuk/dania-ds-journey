```python
import matplotlib.pyplot as plt
import pandas as pd
```

## Questions 1 - 3

```python
data = pd.read_csv("Sleep_health_and_lifestyle_dataset.csv")
```

```python
fig, ax = plt.subplots()
ax.hist(data['Sleep Duration'])
ax.set_title('Sleep duration (hours)')
plt.show()
```

Most people's sleep hours fall in 6-8 hours range

## Questions 4 - 6

```python
counts = data['BMI Category'].value_counts()
counts
```

```python
fig, ax = plt.subplots()
ax.bar(counts.index, counts.values)
ax.set_title('People per BMI category')
plt.show()
```

There is inconsistent naming of BMI categories

## Questions 7 - 9

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
ax1.hist(data['Sleep Duration'])
ax1.set_title('Sleep duration (hours)')
ax2.bar(counts.index, counts.values)
ax2.set_title('People per BMI category')
plt.show()
```

Because both charts are displayed in one Figure, making it easier to compare sleep duration and BMI category distributions at a glance.
