## Question 4 - 6

```python
bjj.iloc[1]
```

```python
bjj.loc[1]
```

```python
sorted_bjj = bjj.sort_values('minutes')
sorted_bjj
```

```python
sorted_bjj.iloc[1] # position 1
```

```python
bjj.loc[1] # label 1
```

```python
# Now that we sorted values based upon minutes, monday is index 1, while Wednesday has column value 1 assigned to it.
# In simple words, iloc tells what is located, while loc tells where does something lies.
```
