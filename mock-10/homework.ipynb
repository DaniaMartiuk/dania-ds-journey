## Questions 7 - 9

```python
# Goal: answer my question — average minutes per belt, most first
# I expect 5 rows back, because there are 5 different belts in my data

answer = (bjj_clean
.groupby('belt')['minutes']

.agg(['mean', 'count'])

.sort_values('mean', ascending=False))
```

```python
print(answer.shape)
```

```python
print(answer)
```
