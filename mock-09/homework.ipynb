## Questions 1 - 3

```python
bjj = pd.DataFrame({
'member_id': [1, 2, 3,4,5,6,7,8,9],
'belt': ['blue', 'white', 'blue', 'purple','red','white','purple','black','blue'],
'minutes': [90, 60, 90, 120, 180, 60,45,200,300],
'rolls': [5, 3, 6, 8, 12, 3, 5,4,20],
})
bjj
```

Groups bjj dataframe by belt, then takes minutes column, applies aggregation mean and count and resets indexes so they would still have belts as a column instead of an index

```python
result = bjj.groupby('belt')['minutes'].agg(['mean', 'count']).reset_index()
result
```

```python
result.sort_values('mean', ascending=False)
```

```python
result
```

Unchanged — sorting handed a copy of result.
It is occurring because sort_values() hands back a copy of a variable instead of interacting with a variable itself
