## Questions 1 - 3

```python
s = pd.Series([90, 60, 25, np.nan, 35, 15])
```

```python
print(s.dropna().mean()) # honest, one row lighter
```

```python
print(s.fillna(0).mean()) # zero is a real value, not a blank
```

```python
print(s.fillna(s.mean()).mean()) # keeps the average, invents certainty
```

## Questions 7 - 9

```python
print(bjj.groupby('belt')['minutes'].agg(['mean', 'count']))
```

Smallest group is purple/red belt with only 1 session, so its average is least trustworthy number in this table

```python
result = bjj.groupby('belt')['minutes'].agg(['mean', 'count']).reset_index()
result
```

.reset_index() allows to bring back belts used for .groupby() back as separate column instead of indexes

```python
print(result.sort_values('mean', ascending=False))
```
